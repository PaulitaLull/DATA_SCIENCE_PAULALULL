## Ejercicio 10
1. Crea una secuencia de números que vayan del 0 al 10.
2. Crea un nuevo array, que parta del anterior, pero sustituya todos los elementos entre el 3 y el 8, por el valor -1
3. Crea un nuevo array que niegue todos los elementos entre el 3 y el 8

````python
sec10=np.arange(11)
print(sec10)
y=sec10.copy()
z=sec10.copy()
# Importante: precedencia de operadores: en Python & se evalúa antes que >= y <=, 
# por lo que necesito poner parentesis para que calcule las comparaciones antes y no de error
y[(y >= 3) & (y <= 8)]=-1
print(y)

# *= quiere decir ese mismo valor por -1
z[(z >= 3) & (z<= 8)] *= -1
print(z)

# Si negar es eliminarlo:
k=y[y!=-1].copy()
print(k)

````

#### Bonus track. Mini hundir la flota
1. Crea un tablero de 10x10 relleno del carácter espacio ( ).
2. Posiciona un par de barcos en [(0,1), (1,1)] y [(1,3), (1,4), (1,5), (1,6)]. Los barcos serán Os mayúsculas. Como ves, un barco de dos posiciones de eslora y otro de cuatro.
3. Recibe un disparo en uno de los barcos, sustituyendo la O por una X
4. Recibe un disparo en agua, sustituyendo uno de los espacios por un guión: -
5. Ignorando los disparos recibidos, ahora prueba a posicionar otro barco de 4 posiciones dentro del tablero, pero **de manera ALEATORIA.**

**Consejo apartado 5**: El barco tendrá una posición inicial aleatoria dentro del tablero, y una orientación también aleatoria (N,S,E,O).

Por ejemplo, si tu posicion inicial fuese (5,0) y la orientación O, no podrias ponerlo porque se sale del tablero. Ahora bien, si fuese orientación este, si podrías, y ocuparía las posiciones [(5,0), (5,1), (5,2), (5,3)]. Por tanto tendrás que iterar con un bucle si es posible encajar el barco en el tablero, y el bucle terminará cuando lo consigas encajar. **Ojo con las posiciones donde ya hay barco!!!**

Usa `import random` para elegir las coordenadas aleatorias.

````python
tablero= np.full((10, 10), " ")
barco1 = [(0, 1), (1, 1)]
barco2 = [(1, 3), (1, 4), (1, 5), (1, 6)]
posiciones_a_cambiar=barco1+barco2


for fila, columna in posiciones_a_cambiar:
    tablero[fila, columna] = "o"
# tablero[1,4]="X"
# tablero[5,5]="-"
print(tablero, "\n")

# Bucle para crear un posible barco nuevo a posicionar Y POSICIONARLO (solo break cuando lo consigue salir)
while True:
    #Creamos var booleana aleatoria: True=horizontal; False=vertical
    longitud_barco=4
    casillas=[]
    horizontal=np.random.rand()<0.5
    if horizontal==True:
        # Creamos una fila aleatoria. Imagina 3
        fila=np.random.randint(0,10)
        #Creamos una col aleatoria. Imagina 8: NOOO, fuerza a que sea de 0 al 7 NO ICLUIDO
        col=np.random.randint(0,7)
        for i in range(longitud_barco):
            casillas.append([fila, col+i])
    else: #Booleano False (orientacion vertical)
        # Creamos una fila aleatoria. Imagina 8: NOOO, fuerza a que sea de 0 a 7 NO ICLUIDO
        col=np.random.randint(0,10)
        # Creamos una col aleatoria.
        fila=np.random.randint(0,7)
        for i in range(longitud_barco):
            casillas.append([fila+i, col])

    # El barco que he creado cabe? Reviso cada una de las posiciones
    # Si cabe en todas las posiciones, libre=True y que lo meta y al terminar salga del bucle, 
    # si no cabe: libre= false, no llega a añadir nada y VUELVE A INTENTAR EL BUCLE WHILE
    # solo sale de while si se llegan a poner las 4casillas
    libre=True
    for j,k in casillas:
        if tablero[j,k]=="o" or tablero[j,k]=="X":
         libre=False
         print("Encontre una posición que no cabe, volvamos a empezar")
         break

    if libre: #true para todas las posiciones de casillas
        for fila, columna in casillas:
            tablero[fila, columna] = "o"
        break

print(tablero)

#Ahora vamos colocar o en casillas y EVITAR SOLAPAMIENTOS
#Hay que comprobar que las 4 posiciones estan libres, y sino volver a sortear

````
