## Ejercicio 8
Concatena los siguientes arrays verticamente

```Python
a = np.arange(10).reshape(2,-1)

```

````python
a = np.arange(10).reshape(2,-1)
print(a)
b = np.repeat(1, 10).reshape(2,-1)
print(b, "\n")

c=np.concatenate([a,b])
print(c)
````

## Ejercicio 10
Encuentra todos los elementos en común entre ambos arrays. [Para ello usa el método `intersect1d`](https://numpy.org/doc/stable/reference/generated/numpy.intersect1d.html)

```Python
a = np.array([1,2,3,2,3,4,3,4,5,6])
b = np.array([7,2,10,2,7,4,9,4,9,8])
```

````python
a = np.array([1,2,3,2,3,4,3,4,5,9])
b = np.array([7,2,10,2,7,4,9,4,9,8])

# Te devuelve que valores se repiten en dos matrices, independientemente de la posicion
# retun_indices true te devuelve en que indices estan los valores repetidos para cada matriz (solo la primera vez que se repite el valor, ya que no cuenta cuantas veces se repite)
intersection= np.intersect1d(a, b, assume_unique=False, return_indices=True)
print(intersection)
````
