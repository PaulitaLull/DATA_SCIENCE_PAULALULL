```python
import requests
import pandas as pd
import json
import os

pd.set_option('max_colwidth', None)

from dotenv import load_dotenv

load_dotenv() # cargo el token
token = os.getenv("MDM_TOKEN") # me la guardo en var
```

```python
#token="MDM_TOKEN"  # redundante, solo sirve para comprobar con luego un if token
url="http://api.mediastack.com/v1/news"

params={
        "access_key": token,   # obligatori
        # "sources": "antena3,lasextanoticias",      # es de pago
        "countries" :"ar, mx",
        "languages":"es",
        # "date":"2001-02-07",
        "limit": 25 , # es de pago. Sino pagas solo te saca 25
        "offset": 100 # a partir de que noticia quiero que me pase. En un bucle poneindo offset 25, podría ir sacando de 25 en 25
    }

response=requests.get(url,params=params)
print(response.status_code)
print(response.json())

json.loads(response.content)


```

```python
data= response.json()
data_results=data["data"]
df=pd.DataFrame(data_results)
df.to_csv("noticias.csv", index=False)
df.head()

df_pedido = df[["author", "title", "description", "url", "source", "country", "published_at"]]
df_pedido.columns=["autor", "titulo", "descripcion", "url","fuente", "pais", "fechapublicacion"]
df_pedido.head()
```
