## Ejercicio 1
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana. La función constará de un único argumento numérico y una salida de tipo string

## Ejercicio 3
Escibe una función que compare dos números. La función tiene dos argumentos y hay tres salidas posibles: que sean iguales, que el primero se  mayor que el segundo, o que el segundo sea mayor que el primero

````python
def mifuncion2(arg1, arg2):
    if arg1==arg2:
        salida="Iguales"
    if arg1<arg2:
        salida="Primero menor del segundo"
    if arg1>arg2:
        salida="Segundo menor del primero"
    return salida

print(mifuncion2(3,5))
````

Escribe una función que tenga un único argumento, un string. La salida de la función tiene que ser un diccionario con el conteo de todas las letras de ese string.

````python
diccionariomio={}
def diccionario(palabra):
    for num, letra in enumerate(palabra):
      diccionariomio[letra]=palabra.count(letra)
    return diccionariomio

palabra="paulala"
diccionario(palabra)


````

````python
def contar_letras(texto):
    conteo = {}
    for letra in texto:
        conteo[letra] = conteo.get(letra, 0) + 1  # cuantas veces llevo contada esta letra?. Si aun no ha salido es 0
    return conteo

print(contar_letras("banana"))
````

## Ejercicio 6
Escribir una función que añada o elimine elementos en una lista. La función necesita los siguientes argumentos:
* lista: la lista donde se añadirán o eliminarán los elementos
* comando: "add" o "remove"
* elemento: Por defecto es None.

````python
lista6=[3,5,6,2]

def editable(lista, comando, elemento):
    lista_interna=lista6.copy()
    if comando=="add":
          lista_interna.append(elemento)
    elif comando=="remove":
          lista_interna.remove(elemento)
    else:
         print("Comando no válido. Usa 'add' o 'remove' de un dato existente.")
    return print(lista_interna)

editable(lista6, "remove", 5)
print(lista6)
````

## Ejercicio 8
Escribe un programa que obtenga el enésimo número de la [serie de Fibonacci](https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci). Tienes que crear una función recursiva con un único argumento.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

````python
def fibo(fn):
    if fn==0:
        return 0
    elif fn==1:
        return 1
    else:
        return fibo(fn-1) + fibo(fn-2)

print(fibo(8))
````

* Función que calcule el área de un cuadrado
* Función que calcule el area de un triángulo

* Dos círculos de radio 10 + un triángulo de base 3 y altura 7
* Un cuadrado de lado = 10 + 3 círculos (uno de radio = 4 y los otros dos de radio = 6) + 5 triángulos de base = 2 + altura = 4

````python
import math

def area_cuadrado(a):
    return a*a

def area_triangulo(a,b):
    return float(a)*float(b)/2

def area_circulo(r):
    return math.pi*r**2


area_figura1=2*area_circulo(10)+area_triangulo(3,7)
area_figura2=area_cuadrado(10+area_circulo(4)+2*area_circulo(6))+5*area_triangulo(2,4)
print(area_figura1)
print(area_figura2)
````
