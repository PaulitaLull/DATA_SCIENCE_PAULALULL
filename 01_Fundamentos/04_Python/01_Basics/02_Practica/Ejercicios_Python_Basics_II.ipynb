## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

radianes = grados*(pi/180)

````python
import math
grados=input("Dime los grados a convertir:")
radianes=float(grados)*(math.pi/180)
print(f"{grados} grados son {radianes} radianes")
````

## Ejercicio 6
Mismo programa que en 5, pero en esta ocasión tienen que ser tres inputs y dos salidas. Una de las salidas que nos indique si todos son iguales, y la otra si al menos dos inputs sí que lo son

````python
a=input("Dime un primer numero al azar:")
b=input("Dime un segundo numero al azar:")
c=input("Dime un tercer numero al azar y te diré si todos los valores que me dijiste son iguales y si al menos dos de ellos lo son")
sol1=a==b==c #booleano tipo true o false
sol2=(a==b or b==c or c==a)
print(f"¿Coinciden los tres valores que me diste al azar?: {sol1}")
print(f"¿Coinciden al menos dos de los valores que me diste al azar?: {sol2}")
````

## Ejercicio 7
Crea un programa que recoja dos inputs. Tiene que comprobar si su suma es igual, superior o inferior a 10

````python
a=input("Dime un primer numero al azar:")
b=input("Dime un segundo numero al azar y te diré si su suma es igual, superior o inferior a 10")
sol=float(a)+float(b)
print(sol)
sol1=sol==10 #booleano tipo true o false
sol2=sol>10
sol3=sol<10
print(f"La suma de los valores es igual que 10: {sol1}")
print(f"La suma de los valores es superior que 10: {sol2}")
print(f"La suma de los valores es inferior que 10: {sol3}")
````

## Ejercicio 10
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."

1. Pásalo todo a mayúsculas
2. Pásalo todo a minúsculas
3. Solo la primera letra de cada palabra en mayúscula, el resto en minúscula
4. Crea una lista dividiéndolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`
6. Elimina las `a` minúsculas

````python
frase="En un lugar de la Mancha, de cuyo nombre no quiero acordarme."
print(frase.upper())
print(frase.lower())
print(frase.title())
print(frase.split())
print(frase.title().replace(",", ";"))
print(frase.replace("a", ""))
````
